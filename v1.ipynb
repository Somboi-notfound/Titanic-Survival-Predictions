```python
import pandas as pd
import numpy as np
 
train = pd.read_csv('/kaggle/input/competitions/titanic/train.csv')
test = pd.read_csv('/kaggle/input/competitions/titanic/test.csv')
 
# fill missing values with median/mode, encode categoricals as numbers
train['Age'] = train['Age'].fillna(train['Age'].median())
train['Embarked'] = train['Embarked'].fillna(train['Embarked'].mode()[0])
train["Sex"] = train["Sex"].map({"male": 1, "female": 0})
train["Embarked"] = train["Embarked"].map({"S":0, "C":1, "Q":2})
 
# combine SibSp + Parch into one family size feature
train["FamilySize"] = train["SibSp"] + train["Parch"] + 1
 
features = ["Sex", "Fare", "Age", "Pclass", "FamilySize", "Embarked"]
X = train[features]
y = train["Survived"]
 
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
 
# hold out 20% for validation so we can check accuracy honestly
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.2, random_state=0)
 
model = RandomForestClassifier(n_estimators=50, max_depth=20, min_samples_split=5, random_state=0)
model.fit(X_train, y_train)
 
from sklearn.metrics import accuracy_score
predictions = model.predict(X_valid)
print("Validation Accuracy:", accuracy_score(y_valid, predictions))
 
# apply the same cleaning/encoding to test data
test["Age"] = test["Age"].fillna(test["Age"].median())
test["Fare"] = test["Fare"].fillna(test["Fare"].median())
test["Embarked"] = test["Embarked"].fillna(test["Embarked"].mode()[0])
test["Sex"] = test["Sex"].map({"male": 1, "female": 0})
test["Embarked"] = test["Embarked"].map({"S": 0, "C": 1, "Q": 2})
test["FamilySize"] = test["SibSp"] + test["Parch"] + 1
 
X_test = test[features]
test_predictions = model.predict(X_test)
 
submission = pd.DataFrame({"PassengerId": test["PassengerId"], "Survived": test_predictions})
submission.to_csv("submission.csv", index=False)
print("submission.csv created!")
```
